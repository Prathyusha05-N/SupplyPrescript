```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "../data/processed/supply_chain_feature_engineered.csv"
)

print(df.shape)
```

```python
target = "late_delivery_risk"

print(df[target].value_counts())
```

```python
leakage_columns = [
    "delivery_status",
    "days_for_shipping_real",
    "shipping_date_dateorders",
    "shipping_delay_days",
    "order_to_shipping_days",
    "shipping_performance"
]
```

```python
id_columns = [
    "customer_id",
    "order_customer_id",
    "order_id",
    "order_item_cardprod_id",
    "order_item_id",
    "product_card_id",
    "product_category_id",
    "product_image",
    "product_name"
]
```

```python
model_df = df.drop(
    columns=leakage_columns + id_columns,
    errors="ignore"
)

print("Model dataset shape:", model_df.shape)
```

```python
X = model_df.drop(columns=[target])
y = model_df[target]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
print("Training target distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True))
```

```python
import os

os.makedirs("../data/model_ready", exist_ok=True)

X_train.to_csv(
    "../data/model_ready/X_train.csv",
    index=False
)

X_test.to_csv(
    "../data/model_ready/X_test.csv",
    index=False
)

y_train.to_csv(
    "../data/model_ready/y_train.csv",
    index=False
)

y_test.to_csv(
    "../data/model_ready/y_test.csv",
    index=False
)

print("All model-ready files saved successfully.")
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
df = pd.read_csv(
    "../data/processed/supply_chain_feature_engineered.csv"
)

print(df.shape)
print("delay_duration_days" in df.columns)
```

```python
delay_target = df["delay_duration_days"]

delay_y_train = delay_target.loc[X_train.index]
delay_y_test = delay_target.loc[X_test.index]

print("Delay target training:", delay_y_train.shape)
print("Delay target testing :", delay_y_test.shape)
```

```python
delay_y_train.to_csv(
    "../data/model_ready/delay_y_train.csv",
    index=False
)

delay_y_test.to_csv(
    "../data/model_ready/delay_y_test.csv",
    index=False
)

print("Delay-duration target files saved successfully.")
```
