```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "../data/raw/DataCoSupplyChainDataset.csv",
    encoding="latin1"
)

print("Original shape:", df.shape)
```

```python
clean_df = df.copy()

print("Cleaning copy created:", clean_df.shape)
```

```python
columns_to_remove = [
    "Customer Email",
    "Customer Fname",
    "Customer Lname",
    "Customer Password",
    "Customer Street",
    "Customer Zipcode"
]

clean_df = clean_df.drop(
    columns=columns_to_remove,
    errors="ignore"
)

print("Shape after removing customer PII:", clean_df.shape)
```

```python
columns_to_remove = [
    "Product Description",
    "Order Zipcode"
]

clean_df = clean_df.drop(
    columns=columns_to_remove,
    errors="ignore"
)

print("Shape after removing unusable columns:", clean_df.shape)
```

```python
missing = clean_df.isnull().sum()

missing[missing > 0]
```

```python
date_columns = [
    col for col in clean_df.columns
    if "date" in col.lower()
]

print("Date columns:")
for col in date_columns:
    print("-", col)
```

```python
date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

for col in date_columns:
    clean_df[col] = pd.to_datetime(
        clean_df[col],
        errors="coerce"
    )

print(clean_df[date_columns].dtypes)
```

```python
for col in date_columns:
    print(
        col,
        "missing after conversion:",
        clean_df[col].isna().sum()
    )
```

```python
clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
    .str.replace("-", "_", regex=False)
)

print(clean_df.columns.tolist())
```

```python
print("Quantity:")
print(clean_df["order_item_quantity"].describe())

print("\nNegative quantities:")
print((clean_df["order_item_quantity"] < 0).sum())
```

```python
print("Product Price:")
print(clean_df["product_price"].describe())

print("\nNegative product prices:")
print((clean_df["product_price"] < 0).sum())
```

```python
print("Discount:")
print(clean_df["order_item_discount"].describe())

print("\nNegative discounts:")
print((clean_df["order_item_discount"] < 0).sum())
```

```python
print("Scheduled shipping days:")
print(clean_df["days_for_shipment_scheduled"].describe())

print("\nInvalid scheduled days:")
print((clean_df["days_for_shipment_scheduled"] <= 0).sum())
```

```python
print("Actual shipping days:")
print(clean_df["days_for_shipping_real"].describe())

print("\nInvalid actual shipping days:")
print((clean_df["days_for_shipping_real"] <= 0).sum())
```

```python
invalid_dates = (
    clean_df["shipping_date_dateorders"]
    < clean_df["order_date_dateorders"]
)

print("Shipping before order:", invalid_dates.sum())
```

```python
print("Order date range:")
print(clean_df["order_date_dateorders"].min())
print(clean_df["order_date_dateorders"].max())

print("\nShipping date range:")
print(clean_df["shipping_date_dateorders"].min())
print(clean_df["shipping_date_dateorders"].max())
```
